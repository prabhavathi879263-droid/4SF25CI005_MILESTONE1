```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("global_food_delivery_enriched_market_2026.csv")

print(df)
```

```python
print(df.head())
```

```python
print(df.shape)

print(df.columns)

print(df.info())

print(df.describe())
```

```python
print(df.isnull().sum())
```

```python
df.fillna(df.mean(numeric_only=True), inplace=True)
df
```

```python
df["metro_city"] = df["metro_city"].astype("category")

df["cuisine_category"] = df["cuisine_category"].astype("category")

df["is_peak_hour"] = df["is_peak_hour"].astype(bool)
print(df)
```

```python
df["high_rating"] = df["customer_rating"] >= 4

print(df.head())
```

```python
print("Total Orders:", len(df))
```

```python
print("Average Order Value:", df["order_value_usd"].mean())
```

```python
print("Average Rating:", df["customer_rating"].mean())
```

```python
print("Average Delivery Time:", df["total_fulfillment_time_min"].mean())
```

```python
city = df.groupby("metro_city")["order_value_usd"].sum()

print(city)
```

```python
cuisine = df.groupby("cuisine_category")["order_value_usd"].mean()

print(cuisine)
```

```python
city.plot(kind="bar", figsize=(10,5))

plt.title("City Wise Order Value")
plt.xlabel("City")
plt.ylabel("Order Value")

plt.show()
```

```python
cuisine.plot(kind="bar", figsize=(10,5))

plt.title("Cuisine Wise Average Order Value")
plt.xlabel("Cuisine")
plt.ylabel("Average Order Value")

plt.show()
```

```python
plt.hist(df["customer_rating"], bins=10)

plt.title("Customer Rating Distribution")
plt.xlabel("Customer Rating")
plt.ylabel("Number of Orders")

plt.show()
```
